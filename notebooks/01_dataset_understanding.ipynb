```python
# count missing values
missing_values = df.isnull().sum()

# missing percentage
missing_percentage = (missing_values / len(df)) * 100

# create summary table
missing_summary = pd.DataFrame({
    "Missing_Values": missing_values,
    "Missing_Percentage": missing_percentage
})

# show only columns with missing values
missing_summary = missing_summary[
    missing_summary["Missing_Values"] > 0
]

display(missing_summary)
```

```python
# missing values visualization

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))

sns.barplot(
    x=missing_summary.index,
    y=missing_summary["Missing_Percentage"]
)

plt.title("Missing Values Percentage")
plt.xlabel("Columns")
plt.ylabel("Missing Percentage (%)")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
duplicate_count = df.duplicated().sum()

print("Total Duplicate Records:", duplicate_count)

# Calculate duplicate percentage

duplicate_percentage = (
    duplicate_count / len(df)
) * 100

print("Duplicate Percentage:", duplicate_percentage)
```

```python
# Select numerical columns

numeric_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns

# Display minimum and maximum values

for col in numeric_columns:
    print(
        f"{col}: Min = {df[col].min()}, "
        f"Max = {df[col].max()}"
    )
```

```python
# Select numerical columns
numeric_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns

# Outlier analysis using IQR
outlier_results = []

for col in numeric_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    outlier_results.append({
        "Column": col,
        "Lower_Limit": lower_limit,
        "Upper_Limit": upper_limit,
        "Outlier_Count": len(outliers),
        "Outlier_Percentage": (
            len(outliers) / len(df)
        ) * 100
    })

# Create summary
outlier_summary = pd.DataFrame(outlier_results)

display(
    outlier_summary.sort_values(
        by="Outlier_Count",
        ascending=False
    )
)
```

```python
# extreme negative values inspecting

display(
    df.nsmallest(10, "Benefit per order")[
        [
            "Benefit per order",
            "Sales",
            "Order Item Discount",
            "Order Item Profit Ratio",
            "Category Name",
            "Delivery Status"
        ]
    ]
)
```

```python
# Negative Profit records filter

negative_profit_df = df[
    df["Benefit per order"] < 0
].copy()

print("Negative Profit Records:", len(negative_profit_df))

display(
    negative_profit_df[
        [
            "Benefit per order",
            "Sales",
            "Order Item Discount",
            "Order Item Discount Rate",
            "Category Name",
            "Shipping Mode",
            "Delivery Status"
        ]
    ].head(10)
)
```

```python
# Product category-wise negative profit

category_loss = negative_profit_df.groupby(
    "Category Name"
).agg(
    Total_Loss_Orders=("Benefit per order", "count"),
    Average_Benefit=("Benefit per order", "mean")
)

display(
    category_loss.sort_values(
        by="Total_Loss_Orders",
        ascending=False
    ).head(10)
)
```

```python
# discount analysis

discount_analysis = df.groupby(
    pd.cut(
        df["Order Item Discount Rate"],
        bins=[-0.01, 0.10, 0.20, 0.30, 0.40, 1.0]
    ),
    observed=False
).agg(
    Total_Orders=("Benefit per order", "count"),
    Negative_Profit_Orders=(
        "Benefit per order",
        lambda x: (x < 0).sum()
    )
)

discount_analysis["Negative_Profit_Percentage"] = (
    discount_analysis["Negative_Profit_Orders"]
    / discount_analysis["Total_Orders"] * 100
)

display(discount_analysis)
```

```python
# Category-wise Loss Percentage

category_profit_analysis = df.groupby(
    "Category Name"
).agg(
    Total_Orders=("Benefit per order", "count"),

    Negative_Profit_Orders=(
        "Benefit per order",
        lambda x: (x < 0).sum()
    ),

    Average_Profit=("Benefit per order", "mean")
)

category_profit_analysis["Loss_Percentage"] = (
    category_profit_analysis["Negative_Profit_Orders"]
    / category_profit_analysis["Total_Orders"]
) * 100

display(
    category_profit_analysis.sort_values(
        by="Loss_Percentage",
        ascending=False
    )
)
```

```python
# Discount and average profit relationship

discount_profit_analysis = df.groupby(
    pd.cut(
        df["Order Item Discount Rate"],
        bins=[-0.01, 0.10, 0.20, 0.30]
    ),
    observed=False
).agg(
    Total_Orders=("Benefit per order", "count"),
    Average_Profit=("Benefit per order", "mean"),
    Median_Profit=("Benefit per order", "median")
)

display(discount_profit_analysis)
```
