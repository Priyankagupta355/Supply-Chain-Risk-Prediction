```python
# step 1.

import pandas as pd
import numpy as np

df = pd.read_csv(
    "../data/raw/DataCoSupplyChainDataset.csv",
    encoding="latin1"
)

print("Dataset Loaded Successfully!")
print("Dataset Shape: ", df.shape)
```

```python
df_clean = df.copy()

print("Original Dataset: ", df.shape)
print("Cleaning Dataset: ", df_clean.shape)
```

```python
missing_values = df_clean.isnull().sum()

missing_values = missing_values[missing_values > 0]

print("Missing Values in Dataset: ")
print(missing_values)

print("\nTotal Missing Values: ", missing_values.sum())
```

```python
missing_percentage = (
    df_clean.isnull().sum() / len(df_clean)
) * 100

missing_percentage = missing_percentage[
    missing_percentage > 0
].sort_values(ascending=False)

print(missing_percentage)
```

```python
# Step 3: Remove unnecessary columns

# Remove completely empty columns
empty_columns = [
    col for col in df_clean.columns
    if df_clean[col].isnull().all()
]

df_clean = df_clean.drop(columns=empty_columns)

# Exclude Order Zipcode from baseline dataset
df_clean = df_clean.drop(columns=["Order Zipcode"])

print("Removed Empty Columns:", empty_columns)
print("Updated Dataset Shape:", df_clean.shape)
```

```python
# check remaining missing values

print("Remaining Missing Values: ")

print(
    df_clean.isnull().sum()[
        df_clean.isnull().sum() > 0
    ]
)
```

```python
# Step 4: Inspect remaining missing values

print("Column Data Types:")
print(df_clean[["Customer Lname", "Customer Zipcode"]].dtypes)

print("\nRows with Missing Values:")

print(
    df_clean[
        df_clean["Customer Lname"].isnull() |
        df_clean["Customer Zipcode"].isnull()
    ][["Customer Lname", "Customer Zipcode"]].head(15)
)
```

```python
# Step 5: Correct Customer Zipcode datatype

df_clean["Customer Zipcode"] = (
    df_clean["Customer Zipcode"].astype("string")
)

print("Updated Data Types:")
print(df_clean[["Customer Lname", "Customer Zipcode"]].dtypes)

print("\nMissing Values:")
print(df_clean[["Customer Lname", "Customer Zipcode"]].isnull().sum())
```

```python
# Step 5.2 Verify dataset

print("Final Dataset Shape: ", df_clean.shape)

print("Total Missing Values: ", df_clean.isnull().sum().sum())

print("Duplicate Rows: ", df_clean.duplicated().sum())
```

```python
# Duplicate records verification

print("Total Order Item IDs: ", df_clean["Order Item Id"].count())

print("Unique Order Item IDs: ", df_clean["Order Item Id"].nunique())

print(
    "Duplicate Order Item IDs: ",
    df_clean["Order Item Id"].duplicated().sum()
)
```

```python
# step 7. saved cleaned dataset

df_clean.to_csv(
    "../data/processed/supply_chain_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
